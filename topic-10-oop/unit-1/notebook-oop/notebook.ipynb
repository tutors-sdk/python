# Building Classes in Python

In this notebook, we will build classes from scratch, exploring attributes, methods, inheritance, properties, and more. By the end, you will be comfortable designing your own object-oriented programs.

## Your First Class

Let's start with a simple `Dog` class that has attributes and methods.

```python
class Dog:
    def __init__(self, name, breed, age):
        self.name = name
        self.breed = breed
        self.age = age

    def bark(self):
        return f"{self.name} says: Woof!"

    def describe(self):
        return f"{self.name} is a {self.age}-year-old {self.breed}"

# Create instances
buddy = Dog("Buddy", "Golden Retriever", 3)
max_dog = Dog("Max", "German Shepherd", 5)

print(buddy.bark())
print(max_dog.describe())
print(f"\nbuddy.name = {buddy.name}")
print(f"max_dog.age = {max_dog.age}")
print(f"\nType of buddy: {type(buddy)}")
print(f"Is buddy a Dog? {isinstance(buddy, Dog)}")
```

## Instance vs. Class Attributes

Instance attributes are unique to each object. Class attributes are shared by all instances of the class.

```python
import math

class Circle:
    # Class attribute: shared by all instances
    pi = 3.14159

    def __init__(self, radius):
        # Instance attribute: unique per object
        self.radius = radius

    def area(self):
        return Circle.pi * self.radius ** 2

    def circumference(self):
        return 2 * Circle.pi * self.radius

c1 = Circle(5)
c2 = Circle(10)

print(f"c1 radius: {c1.radius}, area: {c1.area():.2f}")
print(f"c2 radius: {c2.radius}, area: {c2.area():.2f}")
print(f"\nSame pi? {c1.pi is c2.pi}")
print(f"Circle.pi = {Circle.pi}")

# Class attribute can be accessed via the class itself
print(f"Circumference of c1: {c1.circumference():.2f}")
```

## Special (Dunder) Methods

Special methods let your objects work with Python's built-in operations like `print()`, `==`, `+`, and `len()`.

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        """Called by str() and print()."""
        return f"({self.x}, {self.y})"

    def __repr__(self):
        """Called by repr() and in the REPL."""
        return f"Point({self.x}, {self.y})"

    def __eq__(self, other):
        """Called by == operator."""
        return isinstance(other, Point) and self.x == other.x and self.y == other.y

    def __add__(self, other):
        """Called by + operator."""
        return Point(self.x + other.x, self.y + other.y)

    def distance_to(self, other):
        return ((self.x - other.x)**2 + (self.y - other.y)**2) ** 0.5

p1 = Point(3, 4)
p2 = Point(1, 2)
p3 = Point(3, 4)

print(f"str: {p1}")
print(f"repr: {repr(p1)}")
print(f"p1 == p3: {p1 == p3}")
print(f"p1 == p2: {p1 == p2}")
print(f"p1 + p2 = {p1 + p2}")
print(f"Distance p1 to p2: {p1.distance_to(p2):.2f}")
```

## Inheritance

Inheritance lets you create new classes that share behavior with existing classes. The child class inherits all methods and attributes from the parent.

```python
class Animal:
    def __init__(self, name, sound):
        self.name = name
        self.sound = sound

    def speak(self):
        return f"{self.name} says {self.sound}!"

    def __str__(self):
        return f"{self.name} the {self.__class__.__name__}"

class Dog(Animal):
    def fetch(self, item):
        return f"{self.name} fetches the {item}!"

class Cat(Animal):
    def purr(self):
        return f"{self.name} purrs contentedly..."

# Create instances
dog = Dog("Buddy", "Woof")
cat = Cat("Whiskers", "Meow")

# Both inherit speak()
print(dog.speak())
print(cat.speak())

# Each has its own methods
print(dog.fetch("ball"))
print(cat.purr())

# Polymorphism: same interface, different types
animals = [dog, cat]
for animal in animals:
    print(f"{animal} -> {animal.speak()}")
```

## Using `super()`

The `super()` function lets you call methods from the parent class, which is essential when overriding `__init__` or other methods.

```python
class Employee:
    def __init__(self, name, employee_id, salary):
        self.name = name
        self.employee_id = employee_id
        self.salary = salary

    def __str__(self):
        return f"{self.name} (ID: {self.employee_id})"

    def get_annual_salary(self):
        return self.salary * 12

class Manager(Employee):
    def __init__(self, name, employee_id, salary, department, team_size):
        super().__init__(name, employee_id, salary)  # call parent __init__
        self.department = department
        self.team_size = team_size

    def __str__(self):
        return f"{self.name} (ID: {self.employee_id}, Dept: {self.department})"

    def get_annual_salary(self):
        # Managers get a 10% bonus
        base = super().get_annual_salary()
        bonus = base * 0.10
        return base + bonus

emp = Employee("Alice", "E001", 5000)
mgr = Manager("Bob", "M001", 7000, "Engineering", 8)

print(emp)
print(f"  Annual salary: ${emp.get_annual_salary():,.2f}")
print(mgr)
print(f"  Annual salary (with bonus): ${mgr.get_annual_salary():,.2f}")
print(f"  Team size: {mgr.team_size}")
```

## Properties

Properties provide controlled access to attributes with validation, computed values, and encapsulation.

```python
class Temperature:
    def __init__(self, celsius):
        self.celsius = celsius  # uses the setter!

    @property
    def celsius(self):
        return self._celsius

    @celsius.setter
    def celsius(self, value):
        if value < -273.15:
            raise ValueError(f"Temperature {value}C is below absolute zero!")
        self._celsius = value

    @property
    def fahrenheit(self):
        """Computed property: derived from celsius."""
        return self._celsius * 9/5 + 32

    @property
    def kelvin(self):
        return self._celsius + 273.15

    def __str__(self):
        return f"{self._celsius:.1f}C / {self.fahrenheit:.1f}F / {self.kelvin:.1f}K"

# Test it
t = Temperature(100)
print(f"Boiling: {t}")

t.celsius = 0
print(f"Freezing: {t}")

t.celsius = 37
print(f"Body temp: {t}")

# Validation in action
try:
    t.celsius = -300
except ValueError as e:
    print(f"\nValidation: {e}")
```

## Class Methods and Static Methods

Class methods receive the class itself as the first argument (`cls`), and static methods receive neither the instance nor the class.

```python
class Date:
    def __init__(self, year, month, day):
        self.year = year
        self.month = month
        self.day = day

    def __str__(self):
        return f"{self.year}-{self.month:02d}-{self.day:02d}"

    @classmethod
    def from_string(cls, date_str):
        """Factory method: create a Date from 'YYYY-MM-DD' string."""
        year, month, day = map(int, date_str.split("-"))
        return cls(year, month, day)

    @classmethod
    def today(cls):
        """Factory method: create a Date for today."""
        from datetime import date
        d = date.today()
        return cls(d.year, d.month, d.day)

    @staticmethod
    def is_valid(year, month, day):
        """Check if date values are valid."""
        if not (1 <= month <= 12):
            return False
        days_in_month = [0, 31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
        if month == 2 and (year % 4 == 0 and (year % 100 != 0 or year % 400 == 0)):
            days_in_month[2] = 29
        return 1 <= day <= days_in_month[month]

d1 = Date(2024, 7, 15)
d2 = Date.from_string("2024-12-25")

print(f"d1: {d1}")
print(f"d2: {d2}")
print(f"\nIs 2024-02-29 valid? {Date.is_valid(2024, 2, 29)}")
print(f"Is 2023-02-29 valid? {Date.is_valid(2023, 2, 29)}")
print(f"Is 2024-13-01 valid? {Date.is_valid(2024, 13, 1)}")
```

## Dataclasses

The `@dataclass` decorator automatically generates `__init__`, `__repr__`, `__eq__`, and more based on type-annotated fields.

```python
from dataclasses import dataclass, field

@dataclass
class Student:
    name: str
    age: int
    grades: list = field(default_factory=list)

    def average(self):
        return sum(self.grades) / len(self.grades) if self.grades else 0.0

    def add_grade(self, grade):
        self.grades.append(grade)
        return self

# Auto-generated __init__ and __repr__
s1 = Student("Alice", 20, [90, 85, 92, 88])
s2 = Student("Bob", 21)
s3 = Student("Alice", 20, [90, 85, 92, 88])

print(s1)  # Auto __repr__
print(s2)

print(f"\nAlice's average: {s1.average():.1f}")

s2.add_grade(75).add_grade(80).add_grade(85)
print(f"Bob's average: {s2.average():.1f}")

# Auto __eq__ compares all fields
print(f"\ns1 == s3: {s1 == s3}")
print(f"s1 == s2: {s1 == s2}")
```

## Composition: Has-A Relationships

Composition models "has-a" relationships. An object contains other objects as attributes, rather than inheriting from them.

```python
class Engine:
    def __init__(self, horsepower, fuel_type):
        self.horsepower = horsepower
        self.fuel_type = fuel_type
        self.running = False

    def start(self):
        self.running = True
        return f"{self.horsepower}hp {self.fuel_type} engine started"

    def stop(self):
        self.running = False
        return "Engine stopped"

class Car:
    def __init__(self, make, model, engine):
        self.make = make
        self.model = model
        self.engine = engine  # Car HAS-AN Engine (composition)
        self.speed = 0

    def start(self):
        msg = self.engine.start()
        return f"{self.make} {self.model}: {msg}"

    def accelerate(self, amount):
        if not self.engine.running:
            return "Start the engine first!"
        self.speed += amount
        return f"Speed: {self.speed} km/h"

    def __str__(self):
        status = "running" if self.engine.running else "off"
        return f"{self.make} {self.model} ({status}, {self.speed} km/h)"

# Build a car with an engine
engine = Engine(200, "electric")
my_car = Car("Tesla", "Model 3", engine)

print(my_car)
print(my_car.start())
print(my_car.accelerate(60))
print(my_car.accelerate(40))
print(my_car)
```

```python
class BankAccount:
    # YOUR CODE HERE
    pass

class SavingsAccount(BankAccount):
    # YOUR CODE HERE
    pass

# Test
account = BankAccount("Alice", 1000)
print(account)
account.deposit(500)
account.withdraw(200)
print(account)

savings = SavingsAccount("Bob", 5000, 0.05)
print(savings)
savings.apply_interest()
print(savings)
savings.withdraw(5000)  # should be rejected
```

```python
# SOLUTION
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if amount <= 0:
            print(f"Error: Deposit amount must be positive, got ${amount:.2f}")
            return
        self.balance += amount
        print(f"Deposited ${amount:.2f}. New balance: ${self.balance:.2f}")

    def withdraw(self, amount):
        if amount <= 0:
            print(f"Error: Withdrawal amount must be positive")
            return
        if amount > self.balance:
            print(f"Error: Insufficient funds. Balance: ${self.balance:.2f}, requested: ${amount:.2f}")
            return
        self.balance -= amount
        print(f"Withdrew ${amount:.2f}. New balance: ${self.balance:.2f}")

    def __str__(self):
        return f"BankAccount({self.owner}: ${self.balance:.2f})"

class SavingsAccount(BankAccount):
    MIN_BALANCE = 100

    def __init__(self, owner, balance=0, interest_rate=0.02):
        super().__init__(owner, balance)
        self.interest_rate = interest_rate

    def apply_interest(self):
        interest = self.balance * self.interest_rate
        self.balance += interest
        print(f"Interest applied: ${interest:.2f}. New balance: ${self.balance:.2f}")

    def withdraw(self, amount):
        if self.balance - amount < self.MIN_BALANCE:
            print(f"Error: Withdrawal would leave balance below ${self.MIN_BALANCE:.2f} minimum")
            return
        super().withdraw(amount)

    def __str__(self):
        return f"SavingsAccount({self.owner}: ${self.balance:.2f}, rate: {self.interest_rate:.0%})"

# Test
account = BankAccount("Alice", 1000)
print(account)
account.deposit(500)
account.withdraw(200)
print(account)

print()
savings = SavingsAccount("Bob", 5000, 0.05)
print(savings)
savings.apply_interest()
print(savings)
savings.withdraw(5000)
```

## Exercise: Shape Hierarchy

Create a class hierarchy for geometric shapes:

**Shape (base class):**
- `name` property
- `area()` method (returns 0, meant to be overridden)
- `perimeter()` method (returns 0, meant to be overridden)
- `__str__` showing name, area, and perimeter

**Rectangle(Shape):** width, height
**Circle(Shape):** radius (use `math.pi`)
**Triangle(Shape):** side_a, side_b, side_c (use Heron's formula for area)

```python
import math

class Shape:
    # YOUR CODE HERE
    pass

class Rectangle(Shape):
    # YOUR CODE HERE
    pass

class Circle(Shape):
    # YOUR CODE HERE
    pass

class Triangle(Shape):
    # YOUR CODE HERE
    pass

# Test
shapes = [
    Rectangle(5, 3),
    Circle(7),
    Triangle(3, 4, 5)
]

for shape in shapes:
    print(shape)
```

```python
# SOLUTION
import math

class Shape:
    def __init__(self, name):
        self._name = name

    @property
    def name(self):
        return self._name

    def area(self):
        return 0

    def perimeter(self):
        return 0

    def __str__(self):
        return f"{self.name}: area = {self.area():.2f}, perimeter = {self.perimeter():.2f}"

class Rectangle(Shape):
    def __init__(self, width, height):
        super().__init__("Rectangle")
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)

class Circle(Shape):
    def __init__(self, radius):
        super().__init__("Circle")
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

    def perimeter(self):
        return 2 * math.pi * self.radius

class Triangle(Shape):
    def __init__(self, side_a, side_b, side_c):
        super().__init__("Triangle")
        self.side_a = side_a
        self.side_b = side_b
        self.side_c = side_c

    def area(self):
        # Heron's formula
        s = self.perimeter() / 2
        return math.sqrt(s * (s - self.side_a) * (s - self.side_b) * (s - self.side_c))

    def perimeter(self):
        return self.side_a + self.side_b + self.side_c

# Test
shapes = [
    Rectangle(5, 3),
    Circle(7),
    Triangle(3, 4, 5)
]

for shape in shapes:
    print(shape)
```
