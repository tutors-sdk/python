# Working with Variables

In this notebook we will explore Python's core data types, practice creating variables, and learn how to convert between types. Each section builds on the last, so work through them in order.

## Creating Variables

Let's create variables of each core type and inspect them with `type()`.

````python
# Integer
age = 25
print(f"age = {age}, type = {type(age).__name__}")

# Float
temperature = 98.6
print(f"temperature = {temperature}, type = {type(temperature).__name__}")

# String
name = "Alice"
print(f"name = {name}, type = {type(name).__name__}")

# Boolean
is_student = True
print(f"is_student = {is_student}, type = {type(is_student).__name__}")

# None
result = None
print(f"result = {result}, type = {type(result).__name__}")
````

## Integer Arbitrary Precision

Unlike many other languages, Python integers have no maximum size. They grow as large as your computer's memory allows.

````python
# A googol - 10 to the power of 100
googol = 10 ** 100
print(f"A googol has {len(str(googol))} digits")
print(googol)

# Even bigger!
print(f"\n2 ** 256 = {2 ** 256}")
````

## Float Precision Gotcha

Floating-point numbers are stored in binary, which cannot represent all decimal fractions exactly. This leads to some surprising results.

````python
# The classic gotcha
result = 0.1 + 0.2
print(f"0.1 + 0.2 = {result}")
print(f"0.1 + 0.2 == 0.3? {result == 0.3}")

# How to handle this properly
import math
print(f"\nUsing math.isclose(): {math.isclose(0.1 + 0.2, 0.3)}")
print(f"Using round():        {round(0.1 + 0.2, 1) == 0.3}")

# The Decimal module for exact arithmetic
from decimal import Decimal
exact = Decimal("0.1") + Decimal("0.2")
print(f"Decimal result:       {exact}")
````

## String Methods

Strings in Python are **immutable** -- every method returns a new string rather than modifying the original. Python has dozens of built-in string methods.

````python
text = "  Hello, Python World!  "

# Case methods
print(f"upper():    '{text.upper()}'")
print(f"lower():    '{text.lower()}'")
print(f"title():    '{text.title()}'")

# Whitespace
print(f"strip():    '{text.strip()}'")
print(f"lstrip():   '{text.lstrip()}'")

# Splitting and joining
words = text.strip().split(", ")
print(f"split():    {words}")
print(f"join():     {' | '.join(words)}")

# Searching
print(f"find('Python'): {text.find('Python')}")
print(f"'Python' in text: {'Python' in text}")
````

````python
# String slicing: string[start:stop:step]
message = "Python Programming"

print(f"First 6 chars:  '{message[0:6]}'")
print(f"From index 7:   '{message[7:]}'")
print(f"Last 5 chars:   '{message[-5:]}'")
print(f"Every 2nd char: '{message[::2]}'")
print(f"Reversed:       '{message[::-1]}'")
````

## Type Conversion

Python lets you convert (cast) values between types using built-in functions. Be aware that not all conversions are valid -- some will raise errors.

````python
# String to numbers
age_str = "25"
age_int = int(age_str)
age_float = float(age_str)
print(f"'{age_str}' -> int: {age_int}, float: {age_float}")

# Number to string
pi = 3.14159
pi_str = str(pi)
print(f"{pi} -> str: '{pi_str}'")

# Float to int (truncates!)
print(f"int(3.9) = {int(3.9)}")
print(f"int(-3.9) = {int(-3.9)}")
print(f"round(3.9) = {round(3.9)}")

# Boolean conversions
print(f"\nbool(1) = {bool(1)}")
print(f"bool(0) = {bool(0)}")
print(f"bool('hello') = {bool('hello')}")
print(f"bool('') = {bool('')}")
print(f"bool(None) = {bool(None)}")
````

## f-strings with Formatting

f-strings support a powerful mini-language for formatting numbers, aligning text, and more.

````python
# Number formatting
pi = 3.14159265
print(f"Pi to 2 decimals: {pi:.2f}")
print(f"Pi to 4 decimals: {pi:.4f}")

# Thousands separator
big_num = 1234567890
print(f"With commas: {big_num:,}")
print(f"Currency: \${big_num:,.2f}")

# Percentage
ratio = 0.856
print(f"Percentage: {ratio:.1%}")

# Padding and alignment
for item, price in [("Apple", 1.50), ("Banana", 0.75), ("Cherry", 3.25)]:
    print(f"{item:<10} \${price:>6.2f}")
````

````python
# SOLUTION

celsius = 37.0

# Convert Celsius to Fahrenheit
fahrenheit = celsius * 9/5 + 32
print(f"{celsius:.1f} C = {fahrenheit:.1f} F")

fahrenheit_input = 212.0

# Convert Fahrenheit to Celsius
celsius_result = (fahrenheit_input - 32) * 5/9
print(f"{fahrenheit_input:.1f} F = {celsius_result:.1f} C")

# Bonus: a conversion table
print("\nConversion Table:")
print(f"{'Celsius':>10} {'Fahrenheit':>12}")
print("-" * 24)
for c in range(0, 101, 10):
    f = c * 9/5 + 32
    print(f"{c:>10} {f:>12.1f}")
````
