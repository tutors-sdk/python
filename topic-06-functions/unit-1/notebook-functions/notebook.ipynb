# Function Workshop

In this notebook, we'll practice defining functions, working with parameters, understanding scope, and using higher-order functions.

## Defining and Calling Functions

The basic pattern: `def`, a name, parameters in parentheses, a colon, and an indented body.

````python
def greet(name):
    """Return a greeting for the given name."""
    return f"Hello, {name}!"

# Call the function
print(greet("Alice"))
print(greet("Bob"))

# A function without a return statement returns None
def say_hello(name):
    print(f"Hi there, {name}!")

result = say_hello("Charlie")
print(f"Return value: {result}")
````

## Parameters: Positional, Keyword, and Default

Python offers flexible ways to pass arguments to functions.

````python
def describe_pet(name, animal_type="dog", trick=None):
    """Describe a pet."""
    description = f"{name} is a {animal_type}"
    if trick:
        description += f" that can {trick}"
    return description

# Positional arguments
print(describe_pet("Rex", "dog", "fetch"))

# Keyword arguments (any order)
print(describe_pet(trick="sit", name="Whiskers", animal_type="cat"))

# Using default values
print(describe_pet("Buddy"))  # animal_type defaults to "dog"

# Mix positional and keyword
print(describe_pet("Nemo", animal_type="fish"))
````

## `*args` and `**kwargs`

Accept a variable number of arguments.

````python
# *args collects extra positional arguments into a tuple
def total(*numbers):
    """Sum any number of arguments."""
    print(f"  Received: {numbers} (type: {type(numbers).__name__})")
    return sum(numbers)

print(f"Total: {total(1, 2, 3)}")
print(f"Total: {total(10, 20, 30, 40, 50)}")

# **kwargs collects extra keyword arguments into a dict
def build_profile(first, last, **info):
    """Build a user profile dictionary."""
    profile = {"first_name": first, "last_name": last}
    profile.update(info)
    return profile

print()
print(build_profile("Ada", "Lovelace", field="mathematics", born=1815))
````

## Multiple Return Values

Return multiple values as a tuple and unpack them at the call site.

````python
def min_max_avg(numbers):
    """Return the minimum, maximum, and average of a list."""
    return min(numbers), max(numbers), sum(numbers) / len(numbers)

data = [23, 45, 12, 67, 34, 89, 5]
lo, hi, avg = min_max_avg(data)

print(f"Data: {data}")
print(f"Min: {lo}")
print(f"Max: {hi}")
print(f"Average: {avg:.1f}")

# You can also capture as a single tuple
result = min_max_avg(data)
print(f"\nAs tuple: {result}")
````

## Scope: Local vs. Global

Variables defined inside a function are local to that function. They don't affect variables outside.

````python
x = "I am global"

def demo_scope():
    x = "I am local"  # This is a different x!
    print(f"Inside function: {x}")

demo_scope()
print(f"Outside function: {x}")

# Functions can READ global variables
greeting = "Hello"

def use_global():
    print(f"{greeting}, World!")  # Reads the global

use_global()

# But modifying requires the 'global' keyword
counter = 0

def increment():
    global counter
    counter += 1

increment()
increment()
print(f"Counter after two increments: {counter}")
````

## Lambda Functions

Small anonymous functions, often used as arguments to other functions.

````python
# Lambda syntax: lambda parameters: expression
double = lambda x: x * 2
print(f"double(5) = {double(5)}")

# Most useful with sorted(), map(), filter()
students = [("Alice", 85), ("Bob", 92), ("Charlie", 78), ("Diana", 95)]

# Sort by score (second element)
by_score = sorted(students, key=lambda s: s[1])
print(f"\nSorted by score: {by_score}")

# Sort by score descending
by_score_desc = sorted(students, key=lambda s: s[1], reverse=True)
print(f"Top students: {by_score_desc}")

# Sort strings by last character
words = ["banana", "apple", "cherry"]
by_last = sorted(words, key=lambda w: w[-1])
print(f"\nSorted by last char: {by_last}")
````

## Higher-Order Functions

Functions that take other functions as arguments or return them.

````python
def apply_twice(func, value):
    """Apply a function to a value, then apply it again to the result."""
    return func(func(value))

def add_three(x):
    return x + 3

def double(x):
    return x * 2

print(f"apply_twice(add_three, 10) = {apply_twice(add_three, 10)}")
print(f"apply_twice(double, 5) = {apply_twice(double, 5)}")
print(f"apply_twice(lambda x: x + 1, 0) = {apply_twice(lambda x: x + 1, 0)}")

# Functions returning functions
def make_multiplier(factor):
    """Return a function that multiplies by factor."""
    def multiplier(x):
        return x * factor
    return multiplier

times_three = make_multiplier(3)
times_ten = make_multiplier(10)

print(f"\ntimes_three(7) = {times_three(7)}")
print(f"times_ten(7) = {times_ten(7)}")
````

## Type Hints and Docstrings

Type hints document expected types. Docstrings explain what the function does.

````python
def calculate_average(scores: list[float]) -> float:
    """Calculate the average of a list of scores.

    Args:
        scores: A list of numeric scores.

    Returns:
        The arithmetic mean of the scores.

    Raises:
        ValueError: If the scores list is empty.
    """
    if not scores:
        raise ValueError("Cannot calculate average of empty list")
    return sum(scores) / len(scores)

# Using the function
result = calculate_average([85.0, 92.5, 78.0, 95.5, 88.0])
print(f"Average: {result:.1f}")

# Accessing the docstring
print(f"\nDocstring:\n{calculate_average.__doc__}")
````
