# Conditional Logic in Python

In this notebook, we'll explore how to make decisions in Python using `if`, `elif`, `else`, ternary expressions, and `match/case`.

## Simple `if` Statement

The `if` statement executes a block of code only when its condition is `True`.

```python
temperature = 35

if temperature > 30:
    print("It's hot outside!")

if temperature > 40:
    print("Extreme heat warning!")

print(f"Current temperature: {temperature}°C")
```

## `if/else` — Choosing Between Two Paths

When you need exactly two branches, use `if/else`.

```python
age = 16

if age >= 18:
    print("You are an adult.")
    print("You can vote!")
else:
    print("You are a minor.")
    print(f"Wait {18 - age} more year(s) to vote.")
```

## `if/elif/else` — Multiple Conditions

Use `elif` (short for "else if") when you have more than two possible outcomes. Conditions are evaluated top to bottom, and only the **first matching** branch executes.

```python
score = 85

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"

print(f"Score: {score} -> Grade: {grade}")

# Try different scores
for s in [95, 72, 58, 100]:
    if s >= 90:
        g = "A"
    elif s >= 80:
        g = "B"
    elif s >= 70:
        g = "C"
    elif s >= 60:
        g = "D"
    else:
        g = "F"
    print(f"Score: {s} -> Grade: {g}")
```

## Truthy and Falsy Values

Python evaluates non-boolean values as `True` or `False` in conditional contexts. Understanding what is "truthy" and "falsy" helps you write cleaner code.

```python
# Falsy values
print("=== Falsy values ===")
print(f"bool(0)     = {bool(0)}")
print(f"bool(0.0)   = {bool(0.0)}")
print(f"bool('')    = {bool('')}")
print(f"bool([])    = {bool([])}")
print(f"bool(())    = {bool(())}")
print(f'bool({{}})    = {bool({})}')
print(f"bool(None)  = {bool(None)}")
print(f"bool(False) = {bool(False)}")

print("\n=== Truthy values ===")
print(f"bool(1)       = {bool(1)}")
print(f"bool(-1)      = {bool(-1)}")
print(f"bool('hello') = {bool('hello')}")
print(f"bool([1, 2])  = {bool([1, 2])}")
print(f"bool(3.14)    = {bool(3.14)}")
```

## Ternary Expressions

Python's ternary operator lets you write compact conditional assignments in a single line.

**Syntax:** `value_if_true if condition else value_if_false`

```python
age = 20
status = "adult" if age >= 18 else "minor"
print(f"Age {age}: {status}")

# Ternary in expressions
number = -5
abs_value = number if number >= 0 else -number
print(f"Absolute value of {number} is {abs_value}")

# Ternary in f-strings
count = 1
print(f"You have {count} item{'s' if count != 1 else ''}.")

count = 5
print(f"You have {count} item{'s' if count != 1 else ''}.")
```

## `match/case` (Python 3.10+)

Structural pattern matching provides an elegant way to match values and destructure data.

```python
# Match HTTP status codes
def describe_status(code):
    match code:
        case 200:
            return "OK"
        case 301:
            return "Moved Permanently"
        case 404:
            return "Not Found"
        case 500:
            return "Internal Server Error"
        case _:
            return f"Unknown status: {code}"

for status in [200, 301, 404, 500, 418]:
    print(f"HTTP {status}: {describe_status(status)}")
```

## Nested vs. Flat Conditionals

Deeply nested conditionals are harder to read. Guard clauses and combined conditions can flatten your code.

```python
# Nested (harder to read)
def can_ride_nested(age, height):
    if age >= 12:
        if height >= 140:
            return "You can ride!"
        else:
            return "Too short to ride."
    else:
        return "Too young to ride."

# Flat (guard clause style - easier to read)
def can_ride_flat(age, height):
    if age < 12:
        return "Too young to ride."
    if height < 140:
        return "Too short to ride."
    return "You can ride!"

# Both produce the same results
test_cases = [(10, 150), (14, 130), (15, 145)]
for age, height in test_cases:
    nested = can_ride_nested(age, height)
    flat = can_ride_flat(age, height)
    print(f"Age {age}, Height {height}cm: {flat}")
    assert nested == flat, "Results should match!"
```

```python
# SOLUTION
def calculate(num1, operator, num2):
    match operator:
        case "+":
            return num1 + num2
        case "-":
            return num1 - num2
        case "*":
            return num1 * num2
        case "/":
            if num2 == 0:
                return "Error: Division by zero"
            return num1 / num2
        case "**":
            return num1 ** num2
        case _:
            return f"Error: Unknown operator '{operator}'"

# Test cases
print(calculate(10, "+", 5))
print(calculate(10, "-", 3))
print(calculate(4, "*", 7))
print(calculate(20, "/", 4))
print(calculate(2, "**", 8))
print(calculate(10, "/", 0))
print(calculate(10, "%", 3))
```
