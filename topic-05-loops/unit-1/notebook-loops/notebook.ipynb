# Loop Patterns in Python

This notebook explores `for` loops, `while` loops, `break`/`continue`, useful built-in functions like `enumerate()` and `zip()`, and the power of comprehensions.

## Basic `for` Loop

Iterate over any sequence. The loop variable takes each value in turn.

````python
fruits = ["apple", "banana", "cherry", "date"]

for fruit in fruits:
    print(f"I like {fruit}!")

print(f"\nTotal fruits: {len(fruits)}")
````

## `range()` Variations

The `range()` function generates sequences of numbers. It accepts one, two, or three arguments.

````python
# range(stop) - 0 to stop-1
print("range(5):", list(range(5)))

# range(start, stop)
print("range(2, 8):", list(range(2, 8)))

# range(start, stop, step)
print("range(0, 20, 3):", list(range(0, 20, 3)))

# Negative step (counting backwards)
print("range(10, 0, -2):", list(range(10, 0, -2)))

# Common pattern: repeat N times
n = 3
for i in range(n):
    print(f"  Iteration {i + 1} of {n}")
````

## `while` Loop

Repeat a block of code as long as a condition remains `True`.

````python
# Countdown
count = 5
while count > 0:
    print(f"T-minus {count}...")
    count -= 1
print("Liftoff!")
````

## `break` and `continue`

- `break` exits the loop immediately
- `continue` skips to the next iteration

````python
# break - find the first number divisible by both 3 and 7
print("Finding first number divisible by 3 and 7:")
for n in range(1, 100):
    if n % 3 == 0 and n % 7 == 0:
        print(f"  Found: {n}")
        break

# continue - print only odd numbers
print("\nOdd numbers from 1-10:")
for n in range(1, 11):
    if n % 2 == 0:
        continue
    print(f"  {n}", end="")
print()
````

## `enumerate()` and `zip()`

`enumerate()` gives you both the index and value. `zip()` lets you iterate over multiple sequences in parallel.

````python
# enumerate - get index + value
colors = ["red", "green", "blue", "yellow"]
print("Colors with indices:")
for i, color in enumerate(colors):
    print(f"  {i}: {color}")

# enumerate with custom start
print("\nStarting from 1:")
for i, color in enumerate(colors, start=1):
    print(f"  {i}. {color}")

# zip - iterate over pairs
names = ["Alice", "Bob", "Charlie"]
scores = [85, 92, 78]
print("\nStudent scores:")
for name, score in zip(names, scores):
    print(f"  {name}: {score}")
````

## Nested Loops

A loop inside another loop. The inner loop runs completely for each iteration of the outer loop.

````python
# Multiplication table (1-5)
print("Multiplication Table (1-5):")
print("    ", end="")
for j in range(1, 6):
    print(f"{j:4}", end="")
print()
print("    " + "-" * 20)

for i in range(1, 6):
    print(f"{i:2} |", end="")
    for j in range(1, 6):
        print(f"{i * j:4}", end="")
    print()
````

## List Comprehensions

A concise way to create lists from existing sequences. They combine a loop and an expression into a single line.

````python
# Squares of 0-9
squares = [x ** 2 for x in range(10)]
print("Squares:", squares)

# Filter: only even numbers
evens = [x for x in range(20) if x % 2 == 0]
print("Evens:", evens)

# Transform + filter: squares of odd numbers
odd_squares = [x ** 2 for x in range(10) if x % 2 != 0]
print("Odd squares:", odd_squares)

# Ternary in comprehension
labels = ["even" if x % 2 == 0 else "odd" for x in range(6)]
print("Labels:", labels)
````

## Dict and Set Comprehensions

The same concise pattern works for building dictionaries and sets.

````python
# Dict comprehension: word -> length
words = ["hello", "world", "python", "code"]
word_lengths = {word: len(word) for word in words}
print("Word lengths:", word_lengths)

# Dict comprehension: number -> square
square_map = {n: n**2 for n in range(1, 6)}
print("Square map:", square_map)

# Set comprehension: unique first letters
names = ["Alice", "Aaron", "Bob", "Bella", "Charlie"]
first_letters = {name[0] for name in names}
print("First letters:", sorted(first_letters))

# Set comprehension: unique remainders
remainders = {n % 3 for n in range(20)}
print("Unique remainders mod 3:", remainders)
````
