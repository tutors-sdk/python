## The `print()` Function

The `print()` function is how Python outputs text to the screen. It is the most fundamental function you will use. Every programmer's first program involves printing "Hello, World!" to the screen.

```python
print("Hello, World!")
```

```python
# print() with multiple arguments - separated by spaces
print("Hello,", "Python!")

# Using escape characters for newlines
print("Line 1\nLine 2")

# Using the sep and end parameters
print("A", "B", "C", sep="-")
print("No newline here", end="... ")
print("continued!")
```

## Comments

Comments are notes in your code that Python ignores. They help explain what your code does, both for yourself and for others reading your code later.

- **Single-line comments** start with `#`
- **Multi-line comments** use triple quotes `"""..."""`

```python
# This is a single-line comment - Python ignores it
print("Comments demo")  # You can also put comments at the end of a line

"""
This is a multi-line comment (docstring).
It can span multiple lines.
Often used to document functions and modules.
"""
print("After the multi-line comment")
```

## Basic Arithmetic

Python can be used as a powerful calculator. It supports all standard arithmetic operations plus a few extras like integer division and exponentiation.

```python
# Addition
print("2 + 3 =", 2 + 3)

# Subtraction
print("10 - 4 =", 10 - 4)

# Multiplication
print("6 * 7 =", 6 * 7)

# Division (always returns a float)
print("10 / 3 =", 10 / 3)

# Integer (floor) division
print("10 // 3 =", 10 // 3)

# Modulo (remainder)
print("10 % 3 =", 10 % 3)

# Exponentiation
print("2 ** 10 =", 2 ** 10)
```

## The `input()` Function

The `input()` function lets your program ask the user for information. It always returns a **string**, so you need to convert the result if you want a number.

> **Note**: In this notebook, we simulate `input()` by assigning values directly to variables, since notebooks do not handle interactive input well.

```python
# Simulating input() - in a real script you would write:
# name = input("What is your name? ")
name = "Alice"  # simulating input

print(f"Hello, {name}!")
print(f"Your name has {len(name)} letters.")
```
