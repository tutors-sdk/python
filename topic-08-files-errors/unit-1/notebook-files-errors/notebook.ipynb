# Working with Files and Errors

In this notebook, we will learn how to read and write files in Python, work with JSON and CSV formats, and handle errors using exception handling.

## Writing to a File

Use `open()` with mode `"w"` inside a `with` statement to write text to a file.

```python
# Writing to a file
with open("example.txt", "w") as f:
    f.write("Hello, Python!\n")
    f.write("This is line 2.\n")
    f.write("File I/O is straightforward.\n")

print("File written successfully!")
```

## Reading from a File

Use mode `"r"` (or no mode, since read is the default) to read file contents.

```python
# Reading the entire file
with open("example.txt", "r") as f:
    content = f.read()

print(content)
```

## Reading Line by Line

You can iterate over a file object directly to process one line at a time. This is memory-efficient for large files.

```python
# Reading line by line
with open("example.txt", "r") as f:
    for i, line in enumerate(f, 1):
        print(f"Line {i}: {line.strip()}")
```

## Working with JSON

The `json` module lets you serialize Python objects to JSON and deserialize JSON back to Python objects.

```python
import json

# Write a Python dictionary to a JSON file
student = {
    "name": "Alice",
    "age": 21,
    "courses": ["Python", "Data Science", "Web Dev"],
    "graduated": False
}

with open("student.json", "w") as f:
    json.dump(student, f, indent=2)

print("JSON written!")

# Read it back
with open("student.json", "r") as f:
    loaded = json.load(f)

print(f"Name: {loaded['name']}")
print(f"Courses: {', '.join(loaded['courses'])}")
print(f"Type: {type(loaded)}")
```

## Working with CSV

The `csv` module provides `csv.writer`, `csv.reader`, and `csv.DictReader` for working with comma-separated values.

```python
import csv

# Write CSV data
headers = ["name", "score", "grade"]
rows = [
    ["Alice", 95, "A"],
    ["Bob", 82, "B"],
    ["Charlie", 78, "C+"],
    ["Diana", 91, "A-"]
]

with open("grades.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(headers)
    writer.writerows(rows)

print("CSV written!")

# Read CSV with DictReader
with open("grades.csv", "r") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(f"{row['name']}: {row['score']} ({row['grade']})")
```

## Exception Handling Basics

Use `try` and `except` to catch and handle errors instead of letting them crash your program.

```python
# Basic try/except
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Error: Cannot divide by zero!")

# Catching the exception object
try:
    number = int("not_a_number")
except ValueError as e:
    print(f"ValueError caught: {e}")

print("\nProgram continues running after handled exceptions!")
```

## The Full Pattern: try / except / else / finally

Python provides a complete exception handling structure. Let's trace the execution flow.

```python
def divide(a, b):
    """Demonstrate try/except/else/finally flow."""
    try:
        print(f"  try: attempting {a} / {b}")
        result = a / b
    except ZeroDivisionError:
        print("  except: caught division by zero!")
        result = None
    except TypeError as e:
        print(f"  except: type error - {e}")
        result = None
    else:
        print(f"  else: division succeeded, result = {result}")
    finally:
        print("  finally: this always runs")
    return result

print("Case 1 - Success:")
print(f"  => {divide(10, 3)}\n")

print("Case 2 - ZeroDivisionError:")
print(f"  => {divide(10, 0)}\n")

print("Case 3 - TypeError:")
print(f"  => {divide(10, 'two')}")
```

## Handling Multiple Exceptions

You can handle different exception types differently, or group them together.

```python
def process_value(data, key):
    """Process a value from a dictionary with multiple error handling."""
    try:
        value = data[key]
        number = int(value)
        result = 100 / number
        return result
    except KeyError:
        return f"Error: key '{key}' not found"
    except ValueError:
        return f"Error: '{data[key]}' is not a valid number"
    except ZeroDivisionError:
        return "Error: value is zero, cannot divide"

data = {"a": "5", "b": "0", "c": "hello", "d": "20"}

for key in ["a", "b", "c", "d", "e"]:
    result = process_value(data, key)
    print(f"Key '{key}': {result}")
```

```python
# SOLUTION
import csv

def process_csv_data(filename):
    """Read a CSV file and return valid (name, score) tuples."""
    results = []
    skipped = 0
    try:
        with open(filename, "r") as f:
            reader = csv.DictReader(f)
            for row in reader:
                try:
                    name = row["name"]
                    score = int(row["score"])
                    results.append((name, score))
                except ValueError:
                    print(f"  Skipping {row.get('name', '?')}: invalid score '{row.get('score', '?')}'")
                    skipped += 1
                except KeyError as e:
                    print(f"  Skipping row: missing column {e}")
                    skipped += 1
    except FileNotFoundError:
        print(f"Error: File '{filename}' not found")
        return []

    print(f"\nProcessed: {len(results)} valid, {skipped} skipped")
    return results

# Create test CSV with some bad data
with open("test_scores.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["name", "score"])
    writer.writerow(["Alice", "95"])
    writer.writerow(["Bob", "not_a_number"])
    writer.writerow(["Charlie", "82"])
    writer.writerow(["Diana", ""])
    writer.writerow(["Eve", "91"])

print("Processing test_scores.csv:")
results = process_csv_data("test_scores.csv")
print(f"Results: {results}")

print("\nProcessing nonexistent file:")
process_csv_data("does_not_exist.csv")
```
