# Exploring Python Modules

Python's standard library provides a wealth of modules for common tasks. In this notebook, we will explore some of the most useful ones and practice different import styles.

## Import Styles

Python offers several ways to import modules. Each has its use case.

```python
# Style 1: Import the whole module
import math
print(f"math.pi = {math.pi}")
print(f"math.sqrt(144) = {math.sqrt(144)}")

# Style 2: Import specific items
from math import pi, e
print(f"\npi = {pi}")
print(f"e = {e}")

# Style 3: Import with alias
import math as m
print(f"\nm.ceil(3.2) = {m.ceil(3.2)}")
print(f"m.floor(3.8) = {m.floor(3.8)}")
```

## The `math` Module

Provides mathematical functions and constants beyond basic arithmetic.

```python
import math

# Constants
print(f"pi = {math.pi}")
print(f"e = {math.e}")
print(f"tau = {math.tau}")
print(f"inf = {math.inf}")

# Functions
print(f"\nsqrt(49) = {math.sqrt(49)}")
print(f"ceil(3.2) = {math.ceil(3.2)}")
print(f"floor(3.8) = {math.floor(3.8)}")
print(f"factorial(6) = {math.factorial(6)}")
print(f"gcd(48, 18) = {math.gcd(48, 18)}")
print(f"log(100, 10) = {math.log(100, 10)}")
print(f"pow(2, 10) = {math.pow(2, 10)}")
```

## The `random` Module

Generate random numbers and make random selections.

```python
import random

# Set seed for reproducible results
random.seed(42)

# Random integer in range [a, b]
print(f"randint(1, 10): {random.randint(1, 10)}")
print(f"randint(1, 10): {random.randint(1, 10)}")

# Random choice from a sequence
colors = ["red", "green", "blue", "yellow", "purple"]
print(f"\nchoice: {random.choice(colors)}")

# Random sample (without replacement)
print(f"sample(3): {random.sample(colors, 3)}")

# Shuffle a list in place
numbers = [1, 2, 3, 4, 5]
random.shuffle(numbers)
print(f"\nshuffled: {numbers}")

# Random float [0.0, 1.0)
print(f"\nrandom(): {random.random():.4f}")
print(f"uniform(1, 100): {random.uniform(1, 100):.2f}")
```

## The `datetime` Module

Work with dates, times, and time differences.

```python
from datetime import datetime, date, timedelta

# Current date and time
now = datetime.now()
print(f"Now: {now}")
print(f"Date: {now.date()}")
print(f"Time: {now.time()}")
print(f"Year: {now.year}, Month: {now.month}, Day: {now.day}")

# Formatting
print(f"\nFormatted: {now.strftime('%A, %B %d, %Y')}")
print(f"ISO format: {now.isoformat()}")

# Date arithmetic with timedelta
tomorrow = date.today() + timedelta(days=1)
next_week = date.today() + timedelta(weeks=1)
last_month = date.today() - timedelta(days=30)

print(f"\nToday: {date.today()}")
print(f"Tomorrow: {tomorrow}")
print(f"Next week: {next_week}")
print(f"30 days ago: {last_month}")

# Difference between dates
birthday = date(2000, 6, 15)
age_days = (date.today() - birthday).days
print(f"\nDays since 2000-06-15: {age_days}")
```

## The `collections` Module

Specialized container data types beyond the built-in `dict`, `list`, `set`, and `tuple`.

```python
from collections import Counter, defaultdict, namedtuple

# Counter - count occurrences
words = "the cat sat on the mat the cat ate the rat".split()
word_counts = Counter(words)
print("Counter:")
print(f"  All: {word_counts}")
print(f"  Most common 3: {word_counts.most_common(3)}")

# defaultdict - dictionary with default values
dd = defaultdict(list)
students = [("Alice", "Math"), ("Bob", "Science"), ("Alice", "Science"), ("Bob", "Math")]
for name, subject in students:
    dd[name].append(subject)
print(f"\ndefaultdict:")
for name, subjects in dd.items():
    print(f"  {name}: {subjects}")

# namedtuple - lightweight class
Point = namedtuple("Point", ["x", "y"])
p = Point(3, 4)
print(f"\nnamedtuple:")
print(f"  Point: {p}")
print(f"  x={p.x}, y={p.y}")
```

## The `pathlib` Module

Object-oriented filesystem paths -- the modern replacement for `os.path`.

```python
from pathlib import Path

# Current and home directories
print(f"Current dir: {Path.cwd()}")
print(f"Home dir: {Path.home()}")

# Build paths with / operator
project = Path.cwd() / "my_project" / "src"
print(f"\nProject path: {project}")

# Path properties
p = Path("/home/user/documents/report.txt")
print(f"\nPath: {p}")
print(f"  name: {p.name}")
print(f"  stem: {p.stem}")
print(f"  suffix: {p.suffix}")
print(f"  parent: {p.parent}")
print(f"  parts: {p.parts}")

# List Python files in current directory
py_files = list(Path.cwd().glob("*.py"))
print(f"\nPython files in cwd: {len(py_files)} found")
```

## The `itertools` Module

Functions for efficient looping and combinatorics.

```python
from itertools import chain, product, combinations, permutations

# chain - combine multiple iterables
list1 = [1, 2, 3]
list2 = ["a", "b", "c"]
print(f"chain: {list(chain(list1, list2))}")

# product - cartesian product
colors = ["red", "blue"]
sizes = ["S", "M", "L"]
print(f"\nproduct:")
for combo in product(colors, sizes):
    print(f"  {combo}")

# combinations - order doesn't matter
items = ["A", "B", "C", "D"]
print(f"\ncombinations(4, 2):")
for c in combinations(items, 2):
    print(f"  {c}")

# permutations - order matters
print(f"\npermutations(['X','Y','Z'], 2):")
for p in permutations(["X", "Y", "Z"], 2):
    print(f"  {p}")
```

```python
# SOLUTION
import math
from collections import Counter

def mean(numbers):
    """Calculate the arithmetic mean."""
    return sum(numbers) / len(numbers)

def median(numbers):
    """Calculate the median value."""
    sorted_nums = sorted(numbers)
    n = len(sorted_nums)
    mid = n // 2
    if n % 2 == 0:
        return (sorted_nums[mid - 1] + sorted_nums[mid]) / 2
    else:
        return sorted_nums[mid]

def mode(numbers):
    """Find the most common value using Counter."""
    counter = Counter(numbers)
    return counter.most_common(1)[0][0]

def std_dev(numbers):
    """Calculate the population standard deviation."""
    avg = mean(numbers)
    variance = sum((x - avg) ** 2 for x in numbers) / len(numbers)
    return math.sqrt(variance)

# Test data
data = [4, 7, 2, 9, 4, 1, 5, 4, 8, 3]
print(f"Data: {data}")
print(f"Mean: {mean(data)}")
print(f"Median: {median(data)}")
print(f"Mode: {mode(data)}")
print(f"Std Dev: {std_dev(data):.4f}")
```

```python
# SOLUTION
import random
import string

def generate_password(length=12, uppercase=True, digits=True, special=True):
    """Generate a random password with specified character requirements."""
    # Start with lowercase letters
    chars = list(string.ascii_lowercase)
    password = [random.choice(string.ascii_lowercase)]

    if uppercase:
        chars.extend(string.ascii_uppercase)
        password.append(random.choice(string.ascii_uppercase))
    if digits:
        chars.extend(string.digits)
        password.append(random.choice(string.digits))
    if special:
        special_chars = "!@#$%^&*()-_=+"
        chars.extend(special_chars)
        password.append(random.choice(special_chars))

    # Fill remaining length with random choices from all allowed chars
    remaining = length - len(password)
    password.extend(random.choices(chars, k=remaining))

    # Shuffle so required chars aren't always at the start
    random.shuffle(password)
    return "".join(password)

random.seed(42)
print(f"Default (12): {generate_password()}")
print(f"No uppercase (8): {generate_password(8, uppercase=False)}")
print(f"No special (16): {generate_password(16, special=False)}")
print(f"All options (20): {generate_password(20)}")
```
